# Coffee Exploration: Exports, Imports, and Production

This notebook adapts the existing five-part Python analysis to the three supplied CSV datasets. It explores their structure, summarizes annual values, checks data quality, and calculates findings from the available years.

**Scope note:** the CSV files do not state a measurement unit. Values below are therefore reported in the datasets’ original units. Calendar-year trade figures and coffee crop-year production figures are not directly interchangeable.

## Part 1 — Load the data

Load the CSV files from the project folder and preview the first five rows of each. The original files are left unchanged.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

DATA_FILES = {
    "Coffee Export": Path("Coffee_export-1.csv"),
    "Coffee Import": Path("Coffee_import-1.csv"),
    "Coffee Production": Path("Coffee_production-1.csv"),
}

missing_files = [str(path) for path in DATA_FILES.values() if not path.exists()]
if missing_files:
    raise FileNotFoundError(
        "Run this notebook from the project folder; missing: " + ", ".join(missing_files)
    )

datasets = {name: pd.read_csv(path) for name, path in DATA_FILES.items()}

for name, df in datasets.items():
    print(f"\n===== {name}: first five rows =====")
    display(df.head())
```

## Part 2 — Explore the datasets

The data is arranged in a wide format: each year (or production crop year) has its own column. The final `Total_...` field is an aggregate, not another year, so it is excluded from the time-series comparisons below.

```python
def annual_columns(df):
    """Return year/crop-year columns, excluding the precomputed total."""
    return [
        col for col in df.columns
        if col != "Country"
        and col != "Coffee type"
        and not str(col).startswith("Total_")
    ]

structure_rows = []
for name, df in datasets.items():
    years = annual_columns(df)
    country_count = df["Country"].astype(str).str.strip().nunique()
    structure_rows.append({
        "Dataset": name,
        "Rows": len(df),
        "Columns": len(df.columns),
        "Countries": country_count,
        "First period": years[0],
        "Last period": years[-1],
        "Missing cells": int(df.isna().sum().sum()),
    })

structure = pd.DataFrame(structure_rows).set_index("Dataset")
display(structure)

for name, df in datasets.items():
    print(f"\n===== {name}: column data types =====")
    display(df.dtypes.rename("Data type").to_frame())
```

## Part 3 — Descriptive statistics

For each dataset, summarize the distribution across annual observations (excluding aggregate-total columns). Also compare each period’s summed value. These are descriptive summaries across the dataset, not per-country averages of a single year.

```python
annual_values = {}
period_totals = {}

for name, df in datasets.items():
    periods = annual_columns(df)
    values = df[periods].apply(pd.to_numeric, errors="coerce")
    annual_values[name] = values
    period_totals[name] = values.sum(axis=0, min_count=1)

    observations = values.stack().dropna()
    print(f"\n===== {name}: annual observation statistics =====")
    display(observations.describe().rename("Value").to_frame())

period_summary = pd.DataFrame({name: totals for name, totals in period_totals.items()})
period_summary.index.name = "Year / crop year"
print("\n===== Sum across countries for each period =====")
display(period_summary)
```

## Part 4 — Data quality assessment

Check missing cells, fully duplicated rows, country labels with leading or trailing spaces, and non-numeric values in annual fields. Import country labels are stripped only in temporary analysis expressions; the source CSV is not modified.

```python
quality_rows = []
for name, df in datasets.items():
    annual = annual_values[name]
    country_text = df["Country"].astype(str)
    quality_rows.append({
        "Dataset": name,
        "Missing cells": int(df.isna().sum().sum()),
        "Duplicate rows": int(df.duplicated().sum()),
        "Country labels with extra spaces": int(country_text.str.strip().ne(country_text).sum()),
        "Non-numeric annual cells": int(annual.isna().sum().sum()),
    })

quality = pd.DataFrame(quality_rows).set_index("Dataset")
display(quality)

print("\nCountry labels that need trimming (if any):")
for name, df in datasets.items():
    labels = df.loc[
        df["Country"].astype(str).str.strip().ne(df["Country"].astype(str)),
        "Country",
    ].drop_duplicates().tolist()
    if labels:
        print(f"{name}: {labels}")
    else:
        print(f"{name}: none")
```

## Part 5 — Findings from the data

The trade datasets cover calendar years **1990–2019**; production covers crop years **1990/91–2019/20**. The figures below are calculated from the annual columns only, so precomputed total columns do not inflate the comparison. Since the files do not identify units, the reported values retain the original dataset units.

```python
def endpoint_change(totals):
    first, last = totals.iloc[0], totals.iloc[-1]
    pct_change = (last / first - 1) * 100 if first else float("nan")
    return first, last, pct_change

finding_rows = []
for name, totals in period_totals.items():
    first, last, pct = endpoint_change(totals)
    values = annual_values[name]
    latest_period = values.columns[-1]
    country_names = datasets[name]["Country"].astype(str).str.strip()
    latest_by_country = values[latest_period].groupby(country_names).sum().sort_values(ascending=False)
    leader = latest_by_country.index[0]
    finding_rows.append({
        "Dataset": name,
        "First period": str(totals.index[0]),
        "First-period sum": first,
        "Latest period": str(totals.index[-1]),
        "Latest-period sum": last,
        "Change (%)": pct,
        "Latest-period leader": leader,
        "Leader value": latest_by_country.iloc[0],
    })

findings = pd.DataFrame(finding_rows).set_index("Dataset")
display(findings.style.format({
    "First-period sum": "{:,.0f}",
    "Latest-period sum": "{:,.0f}",
    "Change (%)": "{:+.1f}%",
    "Leader value": "{:,.0f}",
}))

print("\nTop five countries in the latest available period:")
for name, values in annual_values.items():
    latest_period = values.columns[-1]
    country_names = datasets[name]["Country"].astype(str).str.strip()
    leaders = values[latest_period].groupby(country_names).sum().sort_values(ascending=False).head(5)
    print(f"\n{name} — {latest_period}")
    display(leaders.rename("Value").to_frame())
```

### Interpretation

- Across reporting countries, **coffee exports decreased 31.6%** from **4,840,440,000** in 1990 to **3,312,036,352** in 2019. Viet Nam was the largest exporter in 2019 at **1,644,000,000**, followed by Colombia at **820,320,000**.
- The summed **coffee imports increased 82.0%**, from **4,447,920,000** in 1990 to **8,096,640,000** in 2019. The United States of America was the largest importer in 2019 at **1,851,240,000**, followed by Germany at **1,345,920,000**.
- Summed coffee production increased **77.0%** between crop years 1990/91 (**5,593,800,000**) and 2019/20 (**9,903,180,000**). Brazil led production in 2019/20 at **3,492,660,000**; it also had the largest cumulative sum across the listed crop years (**75,082,980,000**).
- The import file has **28 country labels with surrounding spaces out of 35 rows**. All three files have **zero missing cells and zero fully duplicated rows**. Trimming labels in analysis avoids inconsistent display, but any permanent cleanup should preserve the original source files.

**Cautions:** the country coverage differs among export, import, and production datasets; their sums are not a complete global trade balance. Production uses crop years while trade uses calendar years, and no unit is specified in the files. These data describe reported totals but do not explain the causes of trends.
